```python
import os
import getpass

os.environ["GEMINI_API_KEY"] = getpass.getpass("Enter Gemini API key: ")
```

```python
from crewai import Agent, Task, Crew, LLM

llm = LLM(
    model="gemini/gemini-3.5-flash",
    api_key=os.environ["GEMINI_API_KEY"]
)

agent = Agent(
    role="Topic Explainer",
    goal="Explain technical topics in simple language",
    backstory="You are a helpful teacher who explains concepts to students.",
    llm=llm,
    verbose=True
)

task = Task(
    description=(
        "Explain {topic} in simple language. "
        "Include a definition, three applications, "
        "two advantages, and a short conclusion."
    ),
    expected_output="A clear, structured explanation for students.",
    agent=agent
)

crew = Crew(
    agents=[agent],
    tasks=[task],
    verbose=True
)

topic = input("Enter a topic: ")

result = await crew.kickoff_async(inputs={"topic": topic})

print("\nGENERATED ANSWER\n")
print(result)
```
